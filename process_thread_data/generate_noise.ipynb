```python
def clean_csv_nlp(input_file, output_file):
    try:
        # 1. Đọc file CSV
        # Nếu file bị lệch cột (dòng 2, dòng 4 values), ta dùng on_bad_lines='skip' 
        # để bỏ qua các dòng lỗi cấu trúc ngay từ đầu
        df = pd.read_csv(input_file, header=None, names=['text', 'target'], on_bad_lines='skip', engine='python')
        
        # 2. Loại bỏ các dòng trống (NaN)
        df = df.dropna(how='all') # Xóa dòng trống hoàn toàn
        df = df.dropna(subset=['text', 'target']) # Xóa dòng thiếu 1 trong 2 cột
        
        # 3. Loại bỏ các dòng trùng lặp
        df = df.drop_duplicates()
        
        # 4. Đảm bảo dữ liệu là string và trim khoảng trắng thừa
        df['text'] = df['text'].astype(str).str.strip()
        df['target'] = df['target'].astype(str).str.strip()
        
        # 5. Xuất ra file CSV mới (không header, format: text,target)
        df.to_csv(output_file, index=False, header=False, encoding='utf-8-sig')
        
        print(f"Đã xử lý xong! File sạch được lưu tại: {output_file}")
        print(f"Số lượng dòng còn lại: {len(df)}")

    except Exception as e:
        print(f"Có lỗi xảy ra: {e}")

# Thay đổi tên file của bạn ở đây
input_name = 'final_cleaned_threads.csv' 
output_name ='final_cleaned_threads.csv'

# Chạy hàm clean
# clean_csv_nlp(input_name, output_name)
```

```python
typo_prefix = {
    "tr": ["ch"], "ch": ["tr"],
    "s": ["x"], "x": ["s"],
    "gi": ["d", "r", "z"], "d": ["gi", "r", "z"], "r": ["d", "gi", "z"],
    "l": ["n"], "n": ["l"],
    "v": ["d", "z"] # VD: vui -> dui, zui
}

qwerty_neighbors = {
    'q': ['w', 'a'],
    'w': ['e', 's', 'q'],
    'o': ['p', 'i', 'l'],
    'p': ['o', 'l'],
    'a': ['s', 'q', 'z'],
    'l': ['k', 'o', 'p'],
    'r': ['e', 't', 'f'],
    'c': ['v', 'x', 'd'],
    'u': ['y', 'i', 'j']
}

vietnamese_diacritics = set("áàảãạấầẩẫậâắằẳẵặăéèẻẽẹếềểễệêíìỉĩịóòỏõọốồổỗộôớờởỡợơúùủũụứừửữựưýỳỷỹỵđ")
```

```python
import random
import re
```

```python
def has_diacritic(char):
    return char.lower() in vietnamese_diacritics

def get_initial_consonant(word):
    """Bóc tách chính xác cụm phụ âm đầu tiếng Việt"""
    # Gom các cụm 3 chữ -> 2 chữ -> 1 chữ để bắt chính xác nhất
    match = re.match(r'^(ngh|ng|nh|gh|gi|ch|tr|th|ph|kh|qu|[bcdđghklmnpqrstvx])(.*)', word)
    if match:
        return match.group(1), match.group(2) # Trả về: (Phụ âm đầu, Phần vần còn lại)
    return "", word

def apply_typo(word):
    """Giả lập lỗi chính tả (Đã fix lỗi tham lam greedy match)"""
    w = word.lower()
    initial, rest = get_initial_consonant(w)

    # Chỉ khi phụ âm đầu khớp CHÍNH XÁC 100% với key trong từ điển
    if initial in typo_prefix:
        wrong = random.choice(typo_prefix[initial])
        return wrong + rest

    return word

    # Khám lỗi phụ âm ĐẦU
    for correct, wrongs in typo_prefix.items():
        if w.startswith(correct):
            for wrong in wrongs:
                # Cắt bỏ phần phụ âm đúng ở đầu, đắp phần phụ âm sai vào, giữ nguyên vần
                candidates.append(wrong + w[len(correct):])

    # Nếu tìm thấy cách viết sai, bốc ngẫu nhiên 1 biến thể
    if candidates:
        return random.choice(candidates)

    # Nếu từ không nằm trong diện dễ viết sai, trả về nguyên bản
    return word

def word_to_telex(word):
    """Phân rã từ có dấu thành chuỗi gõ phím thô (Buffer của Unikey)"""
    w = word.lower()
    tone = ""
    if re.search(r'[áấắéếíóốớúứý]', w): tone = 's'
    elif re.search(r'[àầằèềìòồờùừỳ]', w): tone = 'f'
    elif re.search(r'[ảẩẳẻểỉỏổởủửỷ]', w): tone = 'r'
    elif re.search(r'[ãẫẵẽễĩõỗỡũữỹ]', w): tone = 'x'
    elif re.search(r'[ạậặẹệịọộợụựỵ]', w): tone = 'j'

    telex_map = {
        'á':'a', 'à':'a', 'ả':'a', 'ã':'a', 'ạ':'a', 'ấ':'aa', 'ầ':'aa', 'ẩ':'aa', 'ẫ':'aa', 'ậ':'aa', 'â':'aa',
        'ắ':'aw', 'ằ':'aw', 'ẳ':'aw', 'ẵ':'aw', 'ặ':'aw', 'ă':'aw', 'é':'e', 'è':'e', 'ẻ':'e', 'ẽ':'e', 'ẹ':'e',
        'ế':'ee', 'ề':'ee', 'ể':'ee', 'ễ':'ee', 'ệ':'ee', 'ê':'ee', 'í':'i', 'ì':'i', 'ỉ':'i', 'ĩ':'i', 'ị':'i',
        'ó':'o', 'ò':'o', 'ỏ':'o', 'õ':'o', 'ọ':'o', 'ố':'oo', 'ồ':'oo', 'ổ':'oo', 'ỗ':'oo', 'ộ':'oo', 'ô':'oo',
        'ớ':'ow', 'ờ':'ow', 'ở':'ow', 'ỡ':'ow', 'ợ':'ow', 'ơ':'ow', 'ú':'u', 'ù':'u', 'ủ':'u', 'ũ':'u', 'ụ':'u',
        'ứ':'uw', 'ừ':'uw', 'ử':'uw', 'ữ':'uw', 'ự':'uw', 'ư':'uw', 'ý':'y', 'ỳ':'y', 'ỷ':'y', 'ỹ':'y', 'ỵ':'y', 'đ':'dd'
    }
    keystrokes = "".join([telex_map.get(c, c) for c in w]) + tone

    # Phục hồi viết hoa nếu từ gốc viết hoa chữ đầu
    return keystrokes.capitalize() if word.istitle() else keystrokes
```

```python
def apply_keyboard_noise_single(word):
    if len(word) < 2: return word
    current_word = word
    error_type = 'qwerty' if random.random() < 0.60 else 'swap'
    if error_type == 'qwerty':
        if any(has_diacritic(c) for c in current_word): current_word = word_to_telex(current_word)
        indices = [i for i, c in enumerate(current_word) if c.lower() in qwerty_neighbors]
        if not indices: return current_word
        idx = random.choice(indices)
        char_l = current_word[idx].lower()
        repl = 'u' if char_l == 'w' and random.random() < 0.5 else random.choice(qwerty_neighbors.get(char_l, [char_l]))
        res = repl.upper() if current_word[idx].isupper() else repl
        return current_word[:idx] + res + current_word[idx+1:]
    elif error_type == 'swap':
        if any(has_diacritic(c) for c in current_word): current_word = word_to_telex(current_word)
        if len(current_word) >= 2:
            idx = random.randint(0, len(current_word) - 2)
            return current_word[:idx] + current_word[idx+1] + current_word[idx] + current_word[idx+2:]
    return current_word
```

```python
def stretch_word(word):
    """Kéo dài ký tự cuối của từ (ví dụ: nha -> nhaaa)"""
    if len(word) > 0:
        return word + word[-1] * random.randint(1, 3)
    return word
```

```python
MAX_NGRAM = max([len(k.split()) for k in tc_dict.keys()]) if tc_dict else 1
```

```python
def process_sentence(sentence):
    if not sentence or not isinstance(sentence, str):
        return "", {'tc': 0, 'typo': 0, 'kb': 0, 'str': 0}

    words = sentence.split()
    if not words: return "", {'tc': 0, 'typo': 0, 'kb': 0, 'str': 0}
    N_orig = len(words)

    # 1. Bóc tách
    raw = []
    for w in words:
        m = re.match(r"^([^\W_]+)(.*)$", w, re.UNICODE)
        if m:
            raw.append({'core': m.group(1).lower(), 'punct': m.group(2), 'title': m.group(1).istitle(), 'upper': m.group(1).isupper(), 'valid': True, 'original': w})
        else:
            raw.append({'valid': False, 'original': w})

    # 2. Tokenization
    tokens, i = [], 0
    while i < len(raw):
        if not raw[i]['valid']:
            tokens.append(raw[i]); i += 1
        else:
            found = False
            for n in range(MAX_NGRAM, 0, -1):
                if i + n <= len(raw) and all(raw[i+k]['valid'] for k in range(n)):
                    comb = " ".join([raw[i+k]['core'] for k in range(n)])
                    if comb in tc_dict:
                        tokens.append({'core': comb, 'punct': raw[i+n-1]['punct'], 'title': raw[i]['title'], 'upper': raw[i]['upper'], 'valid': True, 'original': " ".join([raw[i+k]['original'] for k in range(n)])})
                        i += n; found = True; break
            if not found:
                tokens.append(raw[i]); i += 1

    N = len(tokens)
    tc_c, typo_c, kb_c = [], [], []
    for idx, t in enumerate(tokens):
        if not t['valid']: continue
        core = t['core']
        if core in tc_dict: tc_c.append(idx)
        if get_initial_consonant(core)[0] in typo_prefix: typo_c.append(idx)
        if any(char in qwerty_neighbors for char in core): kb_c.append(idx)

    # 3. PHÂN BỔ QUOTA (LOGIC MỚI - SIẾT VÙNG MIỀN, ĐẨY BÀN PHÍM)
    # A. Teencode: Giữ nguyên
    num_tc = min(math.ceil(len(tc_c) * 0.9), math.ceil(N * 0.25))
    tc_chosen = set(random.sample(tc_c, num_tc)) if tc_c else set()

    # B. Typo: Dùng xác suất câu để ép xuống ~0.7
    typo_chosen = set()
    v_typo = [idx for idx in typo_c if idx not in tc_chosen]
    if v_typo and random.random() < 0.68: # Chỉ 68% số câu có lỗi vùng miền
        typo_chosen = set(random.sample(v_typo, 1)) # Mỗi câu tối đa 1 lỗi

    # C. Bàn phím: Đẩy mạnh lên (Xác suất 18%, Quota max 5)
    kb_chosen = set()
    if N_orig >= 4: # Hạ ngưỡng xuống 4 từ để tăng cơ hội
        c_kb = [idx for idx in kb_c if idx not in tc_chosen and idx not in typo_chosen]
        pot_kb = [idx for idx in c_kb if random.random() < 0.18] # Tăng lên 18%
        if pot_kb:
            # Quota mạnh dạn hơn: dễ dàng ra 2-3 lỗi, thỉnh thoảng 4-5 lỗi
            quota = random.randint(1, 3) if random.random() < 0.7 else random.randint(4, 5)
            kb_chosen = set(random.sample(pot_kb, min(len(pot_kb), quota)))

    # 4. Render
    output = []
    for idx, d in enumerate(tokens):
        if not d['valid']:
            output.append(d['original']); continue

        core = d['core']; new = core
        if idx in tc_chosen:
            vars = tc_dict[core]
            new = random.choices(vars, weights=[1.0/(2**j) for j in range(len(vars))], k=1)[0]
        elif idx in typo_chosen: new = apply_typo(core)
        elif idx in kb_chosen: new = apply_keyboard_noise_single(core)

        if new == core: output.append(d['original'])
        else:
            if d['title']: new = new.capitalize()
            elif d['upper']: new = new.upper()
            output.append(new + d['punct'])

    # 5. Stretch
    stretch_on = 0
    if random.random() < 0.025 and output:
        output[-1] = stretch_word(output[-1]); stretch_on = 1

    return " ".join(output), {'tc': len(tc_chosen), 'typo': len(typo_chosen), 'kb': len(kb_chosen), 'str': stretch_on}
```
